# Graph Renderer Test Cases

This notebook demonstrates small examples and test cases for the draw_graph function.

They aren't explicit software tests that assert failure or success. For the graph drawing function, "Does it look good?" is the key questio.


```python
import matplotlib.pyplot as plt
import numpy as np

from graph_renderer import draw_graph
```

## Test Cases


```python
# Test 1: Single node
fig, ax = plt.subplots(figsize=(2, 2))
draw_graph(
    ax,
    nodes={'A': 0.5},
    positions={'A': (0, 0)},
    node_labels={'A': 'Node A'}
)
plt.title("Test 1: Single node")
plt.show()
```

```python
# Test 2: Dumbbell (two nodes, one edge)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 3))
draw_graph(
    ax1,
    nodes={'A': 0.3, 'B': 0.3},
    positions={'A': (-0.8, 0), 'B': (0.8, 0)},
    edges={('A', 'B'): 4},
)
ax1.set_title("Single directed edge")

# Test 2: Dumbbell (two nodes, two edges)
draw_graph(
    ax2,
    nodes={'A': 0.3, 'B': 0.3},
    positions={'A': (-0.8, 0), 'B': (0.8, 0)},
    edges={('A', 'B'): 2, ('B', 'A'): 4},
)
ax2.set_title("Bidirectional edges")
plt.tight_layout()
plt.show()
```

```python
# Overlapping nodes
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 3))
draw_graph(
    ax1,
    nodes={'A': 0.4, 'B': 0.4},
    positions={'A': (-0.3, 0), 'B': (0.3, 0)},
    edges={('A', 'B'): 2},
)
ax.set_title("Overlapping linked nodes")

draw_graph(
    ax2,
    nodes={'A': 0.2, 'B': 0.2, 'C': 0.2},
    positions={'A': (-0.7, 0), 'B': (0, 0), 'C': (0.7, 0)},
    edges={('A', 'C'): 1},
)
ax.set_title("Node that overlaps link")

plt.tight_layout()
plt.show()
```

```python
# networkx 2-node example with large edge width.
# lack of control over rounded arrowheads was another motivation for rewriting the renderer.
import matplotlib.pyplot as plt
import networkx as nx

G = nx.DiGraph()
G.add_edge('A', 'B', weight=40)

pos = {'A': (-0.2, -0.1), 'B': (0.2, 0.1)}
fig, ax = plt.subplots(figsize=(4, 3))
nx.draw_networkx_edges(G, pos, ax=ax, width=40, arrows=True, edge_color="gray")
ax.axis('off')
plt.show()
```
